## Step 1: Connect to MySQL

```python
import getpass
import mysql.connector
import pandas as pd
from sqlalchemy import create_engine

HOST = "127.0.0.1"
PORT = 3306
USER = "root"
PASSWORD = getpass.getpass("Enter MySQL password: ")

conn = mysql.connector.connect(host=HOST, port=PORT, user=USER, password=PASSWORD)
cursor = conn.cursor()
print("Connected:", conn.is_connected())

engine = create_engine(f"mysql+pymysql://{USER}:{PASSWORD}@{HOST}:{PORT}/superstore_day23")
```

## Step 2: Load Superstore Data

```python
df = pd.read_sql("SELECT * FROM superstore", engine)
df
```

## Step 3: Basic Analysis

```python
pd.read_sql("SELECT Category, SUM(Sales) AS Total_Sales, SUM(Profit) AS Total_Profit FROM superstore GROUP BY Category", engine)
```

```python
pd.read_sql("SELECT Region, AVG(Sales) AS Average_Sales FROM superstore GROUP BY Region", engine)
```

```python
pd.read_sql("SELECT * FROM superstore WHERE Discount >= 0.5", engine)
```

```python
df.groupby('Category')[['Sales','Profit']].sum().sort_values('Sales', ascending=False)
```

## Step 1: Connect to MySQL

```python
import getpass
import mysql.connector
import pandas as pd
from sqlalchemy import create_engine

HOST = "127.0.0.1"
PORT = 3306
USER = "root"
PASSWORD = getpass.getpass("Enter MySQL password: ")

conn = mysql.connector.connect(host=HOST, port=PORT, user=USER, password=PASSWORD)
cursor = conn.cursor()
print("Connected:", conn.is_connected())

engine = create_engine(f"mysql+pymysql://{USER}:{PASSWORD}@{HOST}:{PORT}/superstore_day23")
```

## Step 2: Insert a practice transaction

```python
cursor.execute("""INSERT INTO superstore
(Row_ID, Order_ID, Order_Date, Ship_Date, Ship_Mode, Customer_ID, Customer_Name,
 Segment, Country, City, State, Postal_Code, Region, Product_ID, Category,
 Sub_Category, Product_Name, Sales, Quantity, Discount, Profit)
VALUES
(9999,'TR-2026-0001','2026-09-01','2026-09-03','Second Class','TR-0001','Training Customer',
 'Consumer','United States','Delhi','Delhi',110001,'East','TR-PH-001','Technology',
 'Accessories','Training Wireless Keyboard',299.00,2,0.10,45.00)""")
conn.commit()
print("Practice row inserted")
```

## Step 3: Update the practice transaction

```python
cursor.execute("UPDATE superstore SET Discount = 0.05 WHERE Row_ID = 9999")
conn.commit()
pd.read_sql("SELECT Row_ID, Product_Name, Sales, Discount, Profit FROM superstore WHERE Row_ID = 9999", engine)
```

## Step 4: Delete the practice transaction

```python
cursor.execute("DELETE FROM superstore WHERE Row_ID = 9999")
conn.commit()
print("Practice row deleted")
```

## Step 5: Final Business Summary

```python
pd.read_sql("SELECT Category, SUM(Sales) AS Total_Sales, SUM(Profit) AS Total_Profit FROM superstore GROUP BY Category ORDER BY Total_Sales DESC", engine)
```
